```python
def sjf(queue):
    time = 0
    n = len(queue)
    if n == 0:
        print("No processes scheduled")
        return
    queue.sort(key = lambda x: x[1])
    details = []
    for i in range(n):
        ready = [process for process in queue if process[1] <= time]
        while ready == []:
            time += 1
            ready = [process for process in queue if process[1] <= time]
        ready.sort(key = lambda x: x[2])
        process = ready[0]
        num, AT, BT = process
        if AT > time:
            time = AT
        WT = time - AT
        time += BT
        CT = time
        TAT = time - AT
        details.append((num, AT, BT, CT, WT, TAT))
        queue.remove(process)
    printDetails(details)
    return
```

```python
def makeQueue():
    n = int(input("Enter number of processes: "))
    processes = []
    for i in range(n):
        process_no = i + 1
        print("\nProcess {}:".format(process_no))
        AT = int(input("Enter arrival time: "))
        BT = int(input("Enter burst time: "))
        processes.append((process_no, AT, BT))
    return processes
```

```python
def printDetails(details):
    n = len(details)
    print("\nProcess\tArrival\tBurst\tCompletion")
    print("Number\tTime\tTime\tTime\t")
    for process in details:
        i = 0
        for detail in process:
            print(detail, end = "\t")
            if i == 3:
                break
            i += 1
        print()
    print()

    print("\nProcess\tWaiting\tTurn Around")
    print("Number\tTime\tTime")
    i = 0
    details.sort()
    for process in details:
        for j in [0, 4, 5]:
            print(details[i][j], end = "\t")
        print()
        i += 1
    print()

    AWT = sum([details[i][4] for i in range(n)]) / n
    ATT = sum([details[i][5] for i in range(n)]) / n
    print("Average Waiting Time =", AWT)
    print("Average TurnAround Time =", ATT)
    return
```

```python
queue = makeQueue()
sjf(queue)
```
