```python
def strf(queue):
    time = 0
    num = 0
    entry = 0
    BT = 0
    exit = 0
    status = ''
    current_process = []
    n = len(queue)
    if n == 0:
        print("No processes scheduled")
        return
    processes = queue[:]
    processes.sort()
    queue.sort(key = lambda x: x[1])                                              # Sort processes according to their arrival time
    process_in_processor = False
    details = []
    while queue or process_in_processor:
        ready = [process for process in queue if process[1] <= time]              # Create a ready queue
        while ready == [] and queue:                                              # If no processes are ready, and queue is not empty
            time += 1                                                             # wait for process to arrive
            ready = [process for process in queue if process[1] <= time]
        ready.sort(key = lambda x: x[2])                                          # Sort ready queue according to burst time
        if process_in_processor == True and ready != []:                          # Select process to be executed
            if ready[0][2] < current_process[2]:                                  # Check if current process to be continued or pre-empted
                status = 'Pre-empted'
                ready.append(current_process)
                exit = time
                process_details = (num, entry, BT, exit, status)
                details.append(process_details)
                updated_process = [num, AT, BT, CT, WT, TAT]
                queue.insert(0, updated_process)
                current_process = ready[0]
                ready.remove(current_process)
                queue.remove(current_process)
                entry = time
            else:
                pass
        else:
            if ready != []:
                current_process = ready[0]
                ready.remove(current_process)
                queue.remove(current_process)
                entry = time
        process_in_processor = True
        num, AT, BT, CT, WT, TAT = current_process
        time += 1
        BT -= 1
        if BT == 0:
            status = 'Terminated'
            process_in_processor = False
            CT = time
            exit = time
            TAT = CT - AT
            processes[num-1][3] = CT
            processes[num-1][4] = WT
            processes[num-1][5] = TAT
            process_details = (num, entry, BT, exit, status)
            details.append(process_details)
        current_process = [num, AT, BT, CT, WT, TAT]
        if ready != []:
            for process in queue:                                                     # Increment waiting time of each
                if process[0] in list(zip(*ready))[0]:                                # process in ready queue
                    process[4] += 1
    printDetails(details, processes)
    return
```

```python
def makeQueue():
    n = int(input("Enter number of processes: "))
    process_queue = []
    for i in range(n):
        process_no = i + 1
        print("\nProcess {}:".format(process_no))
        AT = int(input("Enter arrival time: "))
        BT = int(input("Enter burst time: "))
        CT = 0
        WT = 0
        TAT = 0
        process_queue.append([process_no, AT, BT, CT, WT, TAT])
    return process_queue
```

```python
def printDetails(details, processes):
    n = len(processes)

    print("Processor Details: \n")
    print("Process\t\tEntry\t\tRemaining\tExit\t\tStatus")
    print("Number\t\tTime\t\tBurst Time\tTime")
    for detail in details:
        for value in detail:
            print(value, end = "\t\t")
        print()
    print()

    print("Process Details: \n")
    print("Process\t\tArrival\t\tBurst\t\tCompletion\tWaiting\t\tTurn Around")
    print("Number\t\tTime\t\tTime\t\tTime\t\tTime\t\tTime")
    for process in processes:
        for detail in process:
            print(detail, end = "\t\t")
        print()
    print()
    AWT = sum([processes[i][4] for i in range(n)]) / n
    ATT = sum([processes[i][5] for i in range(n)]) / n
    print("Average Waiting Time =", AWT)
    print("Average TurnAround Time =", ATT)
    return
```

```python
queue = makeQueue()
strf(queue)
```
