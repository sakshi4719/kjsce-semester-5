```python
#Fuzzy logic

A = (['x1',0.7], ['x2',0.3], ['x3',0.9], ['x4',0.1])
B = (['x1',0.2], ['x2',0.5], ['x3',0.7], ['x4',0.4])

#Complement of fuzzy set A(x)
# A'(x) = 1 - A(x) for all x ε X

def complement(A):
    updated_value = []
    for i in range(len(A)):
        new_value = 1 - A[i][1]
        temp = A[i][0], round(new_value,3)
        updated_value.append(temp)
    print("The new complement is: \n",updated_value)

complement(A)
complement(B)

#Intersection of fuzzy set
# μ (A∩B)(x)= min{μA(x),μB(x)}

def intersection(A,B):
    interesected = []
    for i in range(len(A)):
        new_value = min(A[i][1], B[i][1])
        temp = A[i][0], round(new_value,3)
        interesected.append(temp)
    print(interesected)

print("\nIntersection of fuzzy set")
intersection(A,B)

#Union of fuzzy set
# μ (A∩B)(x)= max{μA(x),μB(x)}

def union(A,B):
    union = []
    for i in range(len(A)):
        new_value = max(A[i][1], B[i][1])
        temp = A[i][0], round(new_value,3)
        union.append(temp)
    print(union)

print("\nUnion of fuzzy sets")
union(A,B)


#Algebraic product of fuzzy sets
# A(x).B(x) = {(x, μA(x).μB(x)), x ε X }

C = (['x1',0.1], ['x2',0.2], ['x3',0.3], ['x4',0.4])
D = (['x1',0.5], ['x2',0.7], ['x3',0.8], ['x4',0.9])

def product(C,D):
    product = []
    for i in range(len(C)):
        new_value = C[i][1] * D[i][1]
        temp = A[i][0], round(new_value,3)
        product.append(temp)
    print(product)

print("\nProduct of fuzzy sets")
product(C,D)



#Multiplication of fuzzy sets by a crisp number
# A(x).B(x)= {(x, d . μA(x)), x ε X }

C = (['x1',0.1], ['x2',0.2], ['x3',0.3], ['x4',0.4])
d = 0.2

def multiplication(C,d):
    multi = []
    for i in range(len(C)):
        new_value = C[i][1] * d
        temp = A[i][0], round(new_value,3)
        multi.append(temp)
    print(multi)

print("\nProduct of fuzzy sets")
multiplication(C,d)

#Algebraic sum of two fuzzy sets
# A(x)+B(x)= {(x,μA+B(x), x ε X }
# Where μA+B(x) = μA(x)+μB(x) - μA(x).μB(x)

C = (['x1',0.1], ['x2',0.2], ['x3',0.3], ['x4',0.4])
D = (['x1',0.5], ['x2',0.7], ['x3',0.8], ['x4',0.9])

def algebraic_sum(C,D):
    algra_sum = []
    for i in range(len(C)):
        new_value = (C[i][1] + D[i][1]) - (C[i][1] * D[i][1])
        temp = A[i][0], round(new_value,3)
        algra_sum.append(temp)
    print(algra_sum)

print("\nAlgebraic sum of two fuzzy sets")
algebraic_sum(C,D)
```

```python
# Import necessary libraries
import matplotlib.pyplot as plt  # For creating plots
import numpy as np  # For numerical operations
from tabulate import tabulate  # For creating a tabular representation

# Define the average function
def average(a, b):
    c = (a + b) / 2
    return c

# Define a triangular membership function
def triangular_member(x, a, b, c):
    if x <= a:
        return 0.0
    elif a <= x <= b:
        return (x - a) / (b - a)
    elif b <= x <= c:
        return (c - x) / (c - b)
    else:
        return 0.0

# Define a function to calculate and print the membership value
def calculate_and_print_membership(x, a, b):
    c = average(a, b)
    result = triangular_member(x, a, c, b)
    return result

# Define the parameters for the four triangular membership functions
a_values = [0, 5, 10, 15]
b_values = [10, 15, 25, 30]
c_values = [average(a, b) for a, b in zip(a_values, b_values)]

# Create a range of x values for plotting
x_values = np.linspace(0, 30, 100)

# Create plots for each triangular membership function
for a, b, c in zip(a_values, b_values, c_values):
    y_values = [triangular_member(x, a, c, b) for x in x_values]
    plt.plot(x_values, y_values, label=f'a={a}, b={b}')

# Add labels and a legend to the plot
plt.xlabel('X-axis')
plt.ylabel('Membership Value')
plt.title('Triangular Membership Functions')
plt.legend()
plt.show()
print('\n')

# Input values and ranges for which membership values need to be calculated
input_values = [7, 12, 23]
ranges = [(0, 10), (5, 15), (10, 25), (15, 30)]

# Initialize an empty table to store results
table = []

# Loop through the input values and calculate membership values for different ranges
for x in input_values:
    row = [f"x = {x}"]
    for a, b in ranges:
        result = calculate_and_print_membership(x, a, b)
        row.append(result)
    table.append(row)

# Define headers for the table
headers = ["Input Value"] + [f"Range [{a}, {b}]" for a, b in ranges]

# Print the table with headers in a grid format
print(tabulate(table, headers, tablefmt="grid"))
```
