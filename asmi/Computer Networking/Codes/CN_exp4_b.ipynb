```python
def complement(num):
  result = []
  for digit in num:
    result.append(hex(15-int(digit, 16))[2:])
  return "".join(result)
```

```python
def sender(packet_header):
  header = packet_header.replace(" ", "")
  words = [header[i:i+4] for i in range(0, len(header), 4)]
  partial_sum = hex(sum([int(x, 16) for x in words]))[2:]
  partial_sum = "0"*(8-len(partial_sum)) + partial_sum
  carry = [partial_sum[i:i+4] for i in range(0, len(partial_sum), 4)]
  sum_with_carry = hex(sum([int(x, 16) for x in carry]))[2:]
  checksum = complement(sum_with_carry)
  header = header[:-4] + checksum
  return header.upper()
```
