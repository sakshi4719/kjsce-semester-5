```python
def XOR(num1, num2):
  xor = []
  i = 0
  for digit in num1:
      if num2[i] == digit:
          xor.append("0")
      else:
          xor.append("1")
      i += 1
  result = "".join(xor)
  return result
```

```python
def mod2div(data, generator):
    gen_len = len(generator)
    div = data[0:(gen_len)]
    i = 0
    while True:
        nextBit = i + gen_len
        if div[0] == '1':
            div = XOR(div, generator)
        else:
            div = XOR(div, "0" * gen_len)
        if nextBit < len(data):
            div += data[nextBit]
            div = div[1:]
        else:
            break
        i += 1
    crc = div[1:]
    return crc
```

```python
def sender(data, generator):
    print("\nData to be sent:", data)
    gen_len = len(generator)
    padded_data = data + "0" * (gen_len - 1)
    crc = mod2div(padded_data, generator)
    print("Calculated CRC:", crc)
    encrypted_data = data + crc
    print("Encrypted data: ", encrypted_data)
    return
```

```python
def receiver(data, generator):
    print("Received data:", data)
    calculated_crc = mod2div(data, generator)
    print("Calculated CRC:", calculated_crc)
    if int(calculated_crc) == 0:
        print("No error")
        decrypted_data = data[:-3]
    else:
        print("Error detected")
        decrypted_data = "Error"
    return decrypted_data
```

```python
data = input("Enter data to be sent: ")
# data = "10110"
codeGenerator = input("Enter Code Generator: ")
# codeGenerator = "1101"
sender(data, codeGenerator)
recd_data = input("\nEnter received data: ")
decrypted_data = receiver(recd_data, codeGenerator)
print("Decrypted data:", decrypted_data)
```
